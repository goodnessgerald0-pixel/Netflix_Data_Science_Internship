```python
# Task 2 - Exploratory Data Analysis (EDA)
# Step 1: To analyze dataset structure and statistics

import pandas as pd

# to load the cleaned Netflix dataset
df = pd.read_csv(
    r"C:\Users\user\Desktop\internship work\TASK 1\netflix_dataset.csv"
)

# to display dataset structure
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nBasic Statistics:")
print(df.describe())
```

```python
################ STEP 2:  to study content distribution by type

type_counts = df['type'].value_counts()

print("Content Distribution by Type:")
print(type_counts)
```

```python
################### # Step 3: Identify top countries

top_countries = df['country'].value_counts().head(10)

print("Top 10 Countries:")
print(top_countries)



# to split multiple categories in each row
categories = df['listed_in'].str.split(', ')

# to count each category
top_categories = categories.explode().value_counts().head(10)

print("Top 10 Categories:")
print(top_categories)
```

```python
####################### STEP4 : visualizati
import matplotlib.pyplot as plt

# to content distribution by type
type_counts = df['type'].value_counts()

plt.figure(figsize=(7, 5))
type_counts.plot(kind='bar')

plt.title('Netflix Content Distribution by Type')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.xticks(rotation=0)

plt.show()





top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_countries.plot(kind='bar')

plt.title('Top 10 Countries by Number of Netflix Titles', fontsize=16)
plt.xlabel('Country', fontsize=12)
plt.ylabel('Number of Titles', fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()


# Top 10 categories
top_categories = df['listed_in'].str.split(', ').explode().value_counts().head(10)

plt.figure(figsize=(10, 6))
top_categories.plot(kind='bar')

plt.title('Top 10 Netflix Categories', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Number of Titles', fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

### Step 5: Summary of Findings

The Netflix dataset contains 8,790 titles and 14 columns. The analysis shows that Movies are more common than TV Shows, with 6,126 Movies and 2,664 TV Shows.

The United States has the highest number of titles with 3,240, followed by India with 1,057 and the United Kingdom with 638. The most common categories are International Movies, Dramas, and Comedies.

The visualizations clearly show the distribution of Netflix content by type, country, and category. Overall, the dataset contains more Movies than TV Shows and has a large amount of international content.
