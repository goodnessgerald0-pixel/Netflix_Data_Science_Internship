```python

########################### STEP 1  Importortation of the Netflix dataset##############################

import pandas as pd

df = pd.read_csv(r"C:\Users\user\Desktop\internship work\Dataset.csv")

```

```python
# to Check the shape of the dataset
print("Shape:", df.shape)

# to Check column names
print("\nColumns:")
print(df.columns)

# to Check data types
print("\nData types:")
print(df.dtypes)
```

```python
#3#####################################STEP 2: Identify missing values and duplicate records  ###################

# 1. Check missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Check duplicate records
print("Number of duplicate rows:", df.duplicated().sum())

print(df[['Unnamed: 10', 'Unnamed: 11']].head())
```

```python

############################# STEP 3: Remove unnecessary columns ########################

df = df.drop(columns=['Unnamed: 10', 'Unnamed: 11'])

print(df.head())
print(df.columns)
```

```python
# to  Check "Not Given" values
print(df.eq("Not Given").sum())


#################### since still there is not given;

#  to replace "Not Given" with "Unknown"
df['director'] = df['director'].replace('Not Given', 'Unknown')
df['country'] = df['country'].replace('Not Given', 'Unknown')

# Check if "Not Given" still exists
print(df.eq('Not Given').sum())

################# to check format of a date_addded
print(df['date_added'].head(10))
print(df['date_added'].dtype)

#  to convert date_added to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

#to check the result
print(df['date_added'].head(10))
print(df['date_added'].dtype)

print(df['duration'].head(15))
print(df['duration'].dtype)

# to extract the numeric value from duration
df['duration_value'] = df['duration'].str.extract(r'(\d+)')[0].astype(int)

# to extract the unit from duration
df['duration_unit'] = df['duration'].str.extract(r'([A-Za-z]+)')[0]

# to standardize "Seasons" to "Season"
df['duration_unit'] = df['duration_unit'].replace('Seasons', 'Season')

# to check the result
print(df[['duration', 'duration_value', 'duration_unit']].head(15))

```

```python
################STEP4 :Transformation of categorical columns
######## to check values from the type
print(df['type'].value_counts())

# to convert type to categorical data
df['type'] = df['type'].astype('category')

#  to check the data type
print(df['type'].dtype)

######## for rating since its categorical also
print(df['rating'].value_counts())

# to convert rating to categorical data
df['rating'] = df['rating'].astype('category')

# to check the data type
print(df['rating'].dtype)

# to extract year and month from date_added
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month

# to check the result
print(df[['date_added', 'year_added', 'month_added']].head(10))
```

```python
############# STEP 5: to create a clean dataset for further analysis
# final check of the cleaned dataset

print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Save the cleaned dataset
df.to_csv(
    r"C:\Users\user\Desktop\internship work\netflix_dataset.csv",
    index=False
)

print("Clean dataset saved successfully!")
```
